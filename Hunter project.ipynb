```python
import pandas as pd, numpy as np
np.random.seed(101)
n=1025
couriers = ['FedEx','UPS','DHL Express','USPS','Estafeta','Redpack','Paquete Express','OnTrac','LaserShip','FedEx Freight','UPS Ground','UPS Air','USPS Priority','DHL Parcel','FEDEX','ups','dhl','estafeta ','  DHL',None]
products = ['Shea Butter Soap','Charcoal Body Soap','Lavender Scent Spray','Rose Scent Mist','Cocoa Body Lotion','Aloe Body Lotion 500ml','Healing Balm','Lip Balm Strawberry','Vanilla Body Soap Bar','Mango Body Lotion','Jasmine Scent Oil','Menthol Balm']

df=pd.DataFrame({
'Shipment_ID':[f'HNT-{np.random.randint(10000,90000)}' for _ in range(n)],
'Order_Date':np.random.choice(['2023-01-15','15/02/2023','2023/03/10','2023-13-40','01-05-2024',None],n),
'Origin_Warehouse':np.random.choice(['Los Angeles, CA','Dallas TX','Miami FL','Mexico City','Guadalajara MX',None],n),
'Destination_State':np.random.choice(['California','CA','Texas','NY','Florida','Jalisco','CDMX','Nuevo Leon','Yucatan',None],n),
'Destination_Country':np.random.choice(['USA','Mexico'],n),
'Customer_Name':np.random.choice(['Sephora USA','Ulta Beauty','Liverpool MX','Walmart Mexico','Target','Amazon',None],n),
'Product_Category':np.random.choice(['Soap','Body Lotion','Scent','Body Soap','Balm','soap','SOAP'],n),
'Product_Name':np.random.choice(products,n),
'Quantity_Units':np.random.choice([10,50,100,200,-5,None,9999],n),
'Courier':np.random.choice(couriers,n),
'Shipping_Cost_USD':np.random.choice(['$250','300','$-50','USD 450',None,'$'],n),
'Delivery_Status':np.random.choice(['Delivered','In Transit','Delayed','Cancelled',None],n),
'Delay_Days':np.random.choice([0,1,2,5,None,100],n),
})
df=pd.concat([df, df.sample(70, random_state=5)]).sample(frac=1, random_state=42).iloc[:1025]
df.to_csv('hunter_final_all_couriers.csv', index=False)
print("Created -", df.shape)
```

```python
def set_category(name):
    name=str(name).lower()
    if 'soap' in name:
        return 'Body Soap' if 'body' in name else 'Soap'
    if 'lotion' in name:
        return 'Body Lotion'
    if 'scent' in name or 'jasmine' in name:
        return 'Scent'
    if 'balm'in name:
        return "Balm"
    return name
```

```python
df['Product_Category']=df['Product_Name'].apply(set_category)
```
